```python
import os
import pandas as pd

# ============================================================
# CONFIGURAÇÃO
# ============================================================

PASTA = "./Chao"
ARQUIVO_SAIDA = "./Datasets.xlsx"

COLUNAS = [
    "x", "y", "theta",
    "phi_d", "phi_e",
    "phi_d_ref", "phi_e_ref",
    "e_a_d", "e_a_e"
]

# Arquivos que NÃO devem ser incluídos
EXCLUIR = {
    "inf.csv",
    "circ.csv",
    "sin.csv",
    "zzx2-inv.csv"
}

# ============================================================
# CRIA O EXCEL
# ============================================================

with pd.ExcelWriter(
    ARQUIVO_SAIDA,
    engine="openpyxl"
) as writer:

    # Percorre todos os arquivos da pasta
    for arquivo in sorted(os.listdir(PASTA)):

        # Só considera CSV
        if not arquivo.lower().endswith(".csv"):
            continue

        # Verifica se deve ignorar
        if arquivo.lower() in EXCLUIR:
            continue

        caminho = os.path.join(PASTA, arquivo)

        # ----------------------------------------------------
        # Leitura do CSV
        # ----------------------------------------------------
        try:
            # CSV SEM cabeçalho
            df = pd.read_csv(
                caminho,
                header=None
            )

        except:
            df = pd.read_csv(
                caminho,
                sep=";",
                header=None
            )

        # ----------------------------------------------------
        # Verifica quantidade de colunas
        # ----------------------------------------------------
        if df.shape[1] != len(COLUNAS):
            print(
                f"ATENÇÃO: {arquivo} possui {df.shape[1]} "
                f"colunas, mas foram definidas {len(COLUNAS)}."
            )

            print(
                f"    Colunas esperadas: {COLUNAS}"
            )

            continue

        # ----------------------------------------------------
        # Adiciona os nomes das colunas
        # ----------------------------------------------------
        df.columns = COLUNAS

        # ----------------------------------------------------
        # Nome da sheet
        # ----------------------------------------------------
        nome_sheet = os.path.splitext(arquivo)[0]

        # Excel permite no máximo 31 caracteres
        nome_sheet = nome_sheet[:31]

        # ----------------------------------------------------
        # Escreve a sheet
        # ----------------------------------------------------
        df.to_excel(
            writer,
            sheet_name=nome_sheet,
            index=False
        )

        print(
            f"Adicionado: {arquivo} -> {nome_sheet} "
            f"({df.shape[0]} linhas x {df.shape[1]} colunas)"
        )

# ============================================================
# FINAL
# ============================================================

print("\nExcel criado com sucesso:")
print(ARQUIVO_SAIDA)
```
